Tạo DataFrame để quản lý đường dẫn ảnh và labels. DataFrame có 3 cột:
- **path**: đường dẫn đến file ảnh
- **label**: tên lớp (healthy, angular_leaf_spot, bean_rust)

```python
# Đường dẫn đến dữ liệu train và val (local)
train_path = os.path.join(BASE_PATH, 'train')
val_path = os.path.join(BASE_PATH, 'val')

print(f"📂 Train: {train_path}")
print(f"📂 Val: {val_path}")

# Kiểm tra xem thư mục có tồn tại không
if os.path.exists(train_path) and os.path.exists(val_path):
    print("✅ Tìm thấy cả 2 thư mục!")
else:
    print("⚠️  Cảnh báo: Không tìm thấy thư mục train hoặc val")
    print(f"   Hãy đảm bảo có 2 thư mục: {train_path} và {val_path}")
```

## **📂 Cell 2.3: Kiểm tra Thư mục Dữ liệu**

Xác nhận thư mục train và val có tồn tại.

```python
# Định nghĩa từ điển ánh xạ nhãn sang ID số
# Giải thích: Mô hình học máy cần dữ liệu số, không thể xử lý text trực tiếp
# - healthy (khỏe mạnh) -> 0
# - angular_leaf_spot (đốm góc lá) -> 1
# - bean_rust (gỉ sắt đậu) -> 2

label_dict = {"healthy": 0, "angular_leaf_spot": 1, "bean_rust": 2}
```

## **🏷️ Cell 2.4: Định nghĩa Label Mapping**

Tạo dictionary ánh xạ tên lớp sang ID số.

```python
# Hàm tải ảnh và nhãn vào dataframe
# Giải thích chi tiết cách hoạt động:
def load_images(path, label_dict):
    """
    Tải tất cả ảnh từ cấu trúc thư mục và tạo DataFrame
    
    Tham số:
        path: Đường dẫn đến thư mục chính chứa các thư mục con (mỗi thư mục = 1 lớp)
        label_dict: Dictionary ánh xạ tên lớp sang ID số
    
    Trả về:
        DataFrame với 3 cột: path, label, class_id
    """
    df = pd.DataFrame({"path":[], "label":[], "class_id":[]})  # DataFrame rỗng
    
    folder_list = os.listdir(path)  # Lấy danh sách các thư mục con
    
    for folder in folder_list:  # Duyệt qua từng thư mục (mỗi thư mục = 1 lớp)
        img_path = os.path.join(path, folder)  # Tạo đường dẫn đầy đủ
        jpg_list = glob(img_path + '/*.jpg')  # Tìm tất cả file .jpg
        
        for jpg in jpg_list:  # Duyệt qua từng ảnh
            new_data = pd.DataFrame({
                "path": [jpg],  # Đường dẫn ảnh
                "label": [folder],  # Tên lớp
                "class_id": [label_dict[folder]]  # ID số của lớp
            })
            df = pd.concat([df, new_data], ignore_index=True)  # Thêm vào DataFrame
    
    return df
```

## **📥 Cell 2.5: Load Đường dẫn Ảnh**

Duyệt qua các thư mục và thu thập tất cả đường dẫn ảnh cùng labels.

```python
# Tải dữ liệu huấn luyện và kiểm định
# Giải thích: Gọi hàm load_images để tạo DataFrame cho cả 2 tập
train_df = load_images(train_path, label_dict)  # Tập huấn luyện (dùng để train mô hình)

val_df = load_images(val_path, label_dict)  # Tập kiểm định (dùng để đánh giá mô hình)print(f"✅ Đã tải {len(train_df)} ảnh huấn luyện và {len(val_df)} ảnh kiểm định")
```

```python
### **3.1 Advanced Data Augmentation**
# NOTE: If you get a DLL error with torch, run this first in a terminal:
# pip uninstall torch torchvision torchaudio
# pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu

# Import PyTorch transforms (commented out to avoid DLL error - using PIL instead)
# import torch
# import torchvision.transforms as transforms

# For now, we'll use PIL and numpy for augmentation instead
from PIL import Image, ImageEnhance
import numpy as np
import random

# Advanced training transformations with extensive augmentation
def augment_image(img):
    """
    Apply random augmentations to an image using PIL
    Args:
        img: PIL Image object
    Returns:
        Augmented PIL Image
    """
    # Random horizontal flip
    if random.random() > 0.5:
        img = img.transpose(Image.FLIP_LEFT_RIGHT)
    
    # Random rotation (-15 to 15 degrees)
    if random.random() > 0.5:
        angle = random.uniform(-15, 15)
        img = img.rotate(angle, fillcolor=(0, 0, 0))
    
    # Random brightness adjustment
    if random.random() > 0.5:
        enhancer = ImageEnhance.Brightness(img)
        img = enhancer.enhance(random.uniform(0.8, 1.2))
    
    # Random contrast adjustment
    if random.random() > 0.5:
        enhancer = ImageEnhance.Contrast(img)
        img = enhancer.enhance(random.uniform(0.8, 1.2))
    
    return img

print("✅ Image augmentation functions defined (using PIL instead of torch)")
```

```python
# Cài đặt thư viện cần thiết cho feature extraction
# Giải thích: scikit-image cung cấp nhiều công cụ xử lý ảnh mạnh mẽ

# Nếu chưa cài, chạy lệnh sau trong terminal:
# !pip install scikit-image opencv-python

import numpy as np
import cv2
from skimage import color, feature
from skimage.feature import graycomatrix, graycoprops, local_binary_pattern
from scipy import ndimage
from sklearn.preprocessing import StandardScaler

print("✅ Đã import thư viện feature extraction thành công!")
```

```python
def extract_color_features(image_path):
    """
    Trích xuất đặc trưng màu sắc từ ảnh
    
    Đặc trưng bao gồm:
    - Color histogram (RGB): 3 channels x 32 bins = 96 features
    - Color moments (mean, std): 3 channels x 2 = 6 features  
    - HSV histogram: 3 channels x 16 bins = 48 features
    
    Tổng: 150 color features
    """
    # Đọc ảnh
    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB
    img = cv2.resize(img, (224, 224))  # Resize về kích thước chuẩn
    
    features = []
    
    # 1. RGB Color Histogram (96 features)
    for i in range(3):  # Duyệt qua 3 kênh R, G, B
        hist = cv2.calcHist([img], [i], None, [32], [0, 256])
        hist = hist.flatten() / hist.sum()  # Normalize
        features.extend(hist)
    
    # 2. Color Moments - Mean & Std (6 features)
    for i in range(3):
        channel = img[:, :, i]
        features.append(channel.mean())  # Trung bình màu
        features.append(channel.std())   # Độ lệch chuẩn màu
    
    # 3. HSV Color Space (48 features)
    hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
    for i in range(3):  # H, S, V channels
        hist = cv2.calcHist([hsv], [i], None, [16], [0, 256])
        hist = hist.flatten() / hist.sum()
        features.extend(hist)
    
    return np.array(features)


def extract_texture_features(image_path):
    """
    Trích xuất đặc trưng texture từ ảnh
    
    Đặc trưng bao gồm:
    - GLCM features: 5 metrics x 4 angles = 20 features
    - LBP histogram: 26 features
    - Edge density: 1 feature
    
    Tổng: 47 texture features
    """
    # Đọc ảnh và chuyển sang grayscale
    img = cv2.imread(image_path)
    img = cv2.resize(img, (224, 224))
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    features = []
    
    # 1. GLCM Features (20 features)
    # GLCM = Gray Level Co-occurrence Matrix - đo độ tương quan giữa các pixels
    distances = [1, 3]  # Khoảng cách giữa pixels
    angles = [0, np.pi/4, np.pi/2, 3*np.pi/4]  # Các góc 0°, 45°, 90°, 135°
    
    glcm = graycomatrix(gray, distances=distances, angles=angles, 
                        levels=256, symmetric=True, normed=True)
    
    # Tính các metrics từ GLCM
    for prop in ['contrast', 'dissimilarity', 'homogeneity', 'energy', 'correlation']:
        values = graycoprops(glcm, prop)
        features.extend(values.flatten())
    
    # 2. LBP - Local Binary Patterns (26 features)
    # LBP mô tả texture cục bộ xung quanh mỗi pixel
    radius = 3
    n_points = 8 * radius
    lbp = local_binary_pattern(gray, n_points, radius, method='uniform')
    
    # Tạo histogram của LBP
    hist_lbp, _ = np.histogram(lbp, bins=26, range=(0, 26))
    hist_lbp = hist_lbp / hist_lbp.sum()  # Normalize
    features.extend(hist_lbp)
    
    # 3. Edge Density (1 feature)
    # Đo mật độ cạnh trong ảnh - hữu ích cho phát hiện đốm bệnh
    edges = cv2.Canny(gray, 100, 200)
    edge_density = edges.sum() / (edges.shape[0] * edges.shape[1])
    features.append(edge_density)
    
    return np.array(features)


def extract_all_features(image_path):
    """
    Trích xuất TẤT CẢ đặc trưng từ một ảnh
    
    Tổng: 150 (color) + 47 (texture) = 197 features
    """
    color_feat = extract_color_features(image_path)
    texture_feat = extract_texture_features(image_path)
    
    return np.concatenate([color_feat, texture_feat])


print("✅ Đã định nghĩa các hàm trích xuất đặc trưng!")
print("   - extract_color_features(): 150 đặc trưng màu sắc")
print("   - extract_texture_features(): 47 đặc trưng texture")
print("   - extract_all_features(): 197 đặc trưng tổng hợp")
```

```python
# Trích xuất đặc trưng cho TẤT CẢ ảnh trong dataset
# Lưu ý: Quá trình này có thể mất vài phút!

print("🔄 Bắt đầu trích xuất đặc trưng cho toàn bộ dataset...")
print("   Điều này có thể mất 5-10 phút tùy kích thước dữ liệu...\n")

# Trích xuất cho training set
train_features = []
train_labels = []

for idx in range(len(train_df)):
    if idx % 100 == 0:  # In tiến độ mỗi 100 ảnh
        print(f"   Đã xử lý {idx}/{len(train_df)} ảnh training...")
    
    img_path = train_df.iloc[idx]['path']
    label = train_df.iloc[idx]['class_id']
    
    try:
        features = extract_all_features(img_path)
        train_features.append(features)
        train_labels.append(label)
    except Exception as e:
        print(f"   ⚠️ Lỗi với ảnh {img_path}: {e}")
        continue

# Trích xuất cho validation set
val_features = []
val_labels = []

for idx in range(len(val_df)):
    if idx % 20 == 0:  # In tiến độ mỗi 20 ảnh
        print(f"   Đã xử lý {idx}/{len(val_df)} ảnh validation...")
    
    img_path = val_df.iloc[idx]['path']
    label = val_df.iloc[idx]['class_id']
    
    try:
        features = extract_all_features(img_path)
        val_features.append(features)
        val_labels.append(label)
    except Exception as e:
        print(f"   ⚠️ Lỗi với ảnh {img_path}: {e}")
        continue

# Convert to numpy arrays
X_train_features = np.array(train_features)
y_train = np.array(train_labels)
X_val_features = np.array(val_features)
y_val = np.array(val_labels)

# Chuẩn hóa features (quan trọng cho ML algorithms!)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_features)
X_val_scaled = scaler.transform(X_val_features)

print("\n✅ Hoàn thành trích xuất đặc trưng!")
print(f"   Training features shape: {X_train_scaled.shape}")
print(f"   Validation features shape: {X_val_scaled.shape}")
print(f"   Số đặc trưng mỗi ảnh: {X_train_scaled.shape[1]}")

# Lưu features để tái sử dụng (tùy chọn)
np.save('train_features.npy', X_train_scaled)
np.save('val_features.npy', X_val_scaled)
np.save('train_labels.npy', y_train)
np.save('val_labels.npy', y_val)
print("💾 Đã lưu features vào file .npy để tái sử dụng!")
```

```python
# Elbow Method để tìm số cụm tối ưu
print("🔍 Đang chạy Elbow Method để tìm số cụm tối ưu...")

inertias = []  # Within-cluster sum of squares
silhouette_scores = []
K_range = range(2, 11)  # Test từ 2 đến 10 clusters

for k in K_range:
    print(f"   Testing k={k}...", end=" ")
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_train_scaled)
    
    inertias.append(kmeans.inertia_)
    sil_score = silhouette_score(X_train_scaled, labels)
    silhouette_scores.append(sil_score)
    print(f"Silhouette: {sil_score:.3f}")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Elbow Method
axes[0].plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
axes[0].set_xlabel('Số cụm (k)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Inertia (Within-Cluster Sum of Squares)', fontsize=12, fontweight='bold')
axes[0].set_title('Elbow Method for Optimal k', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)
axes[0].axvline(x=3, color='red', linestyle='--', alpha=0.7, label='k=3 (số lớp thực tế)')
axes[0].legend()

# Plot 2: Silhouette Analysis
axes[1].plot(K_range, silhouette_scores, 'ro-', linewidth=2, markersize=8)
axes[1].set_xlabel('Số cụm (k)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Silhouette Score', fontsize=12, fontweight='bold')
axes[1].set_title('Silhouette Analysis for Optimal k', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)
axes[1].axvline(x=3, color='red', linestyle='--', alpha=0.7, label='k=3 (số lớp thực tế)')
axes[1].axhline(y=max(silhouette_scores), color='green', linestyle='--', alpha=0.5)
axes[1].legend()

plt.tight_layout()
plt.show()

# Tìm k tối ưu
optimal_k = K_range[np.argmax(silhouette_scores)]
print(f"\n✅ Số cụm tối ưu theo Silhouette Score: k = {optimal_k}")
print(f"   Silhouette Score cao nhất: {max(silhouette_scores):.4f}")
print(f"   (Lưu ý: k=3 tương ứng với 3 lớp thực tế của dữ liệu)")
```

```python
# Hierarchical Clustering
print("🔍 Áp dụng Hierarchical Clustering...")

# Lấy sample nhỏ để vẽ dendrogram (vì dendrogram với 1000+ samples rất rối)
sample_size = 100
sample_indices = np.random.choice(len(X_train_scaled), sample_size, replace=False)
X_sample = X_train_scaled[sample_indices]
y_sample = y_train[sample_indices]

# Tạo linkage matrix cho dendrogram
Z = linkage(X_sample, method='ward')  # Ward minimizes variance

# Áp dụng Agglomerative Clustering
hierarchical = AgglomerativeClustering(n_clusters=3, linkage='ward')
hier_labels_full = hierarchical.fit_predict(X_train_scaled)
hier_labels_sample = hierarchical.fit_predict(X_sample)

# Đánh giá
sil_hier = silhouette_score(X_train_scaled, hier_labels_full)
ari_hier = adjusted_rand_score(y_train, hier_labels_full)

print(f"\n📊 Kết quả Hierarchical Clustering:")
print(f"   Silhouette Score: {sil_hier:.4f}")
print(f"   Adjusted Rand Index: {ari_hier:.4f}")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Dendrogram (cho sample)
dendrogram(Z, ax=axes[0], color_threshold=0, above_threshold_color='black')
axes[0].set_xlabel('Sample Index', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Distance', fontsize=12, fontweight='bold')
axes[0].set_title(f'Dendrogram (Sample of {sample_size} images)', 
                  fontsize=14, fontweight='bold')
axes[0].axhline(y=Z[-2, 2], color='red', linestyle='--', 
                label=f'Cut at {Z[-2, 2]:.1f} → 3 clusters')
axes[0].legend()

# Plot 2: Hierarchical clusters (full dataset)
scatter2 = axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=hier_labels_full, 
                           cmap='viridis', alpha=0.6, edgecolors='black', linewidth=0.5)
axes[1].set_xlabel('PC1', fontsize=12, fontweight='bold')
axes[1].set_ylabel('PC2', fontsize=12, fontweight='bold')
axes[1].set_title(f'Hierarchical Clustering\nSilhouette: {sil_hier:.3f}', 
                  fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)
plt.colorbar(scatter2, ax=axes[1], label='Cluster ID')

plt.tight_layout()
plt.show()

print(f"\n💡 Đặc điểm Hierarchical Clustering:")
print(f"   ✅ Ưu điểm: Visualize cấu trúc phân cấp, không cần init random")
print(f"   ⚠️ Nhược điểm: Chậm với dữ liệu lớn (O(n²))")
```

```python
# DBSCAN Clustering (Density-Based)
print("🔍 Áp dụng DBSCAN Clustering...")
print("   DBSCAN: Density-Based Spatial Clustering of Applications with Noise")
print("   Ưu điểm: Tự động tìm số clusters, phát hiện outliers\n")

from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors

# Tìm eps tối ưu bằng k-distance graph
# Lấy sample để tính nhanh
sample_size_dbscan = min(1000, len(X_train_scaled))
sample_indices_dbscan = np.random.choice(len(X_train_scaled), sample_size_dbscan, replace=False)
X_sample_dbscan = X_train_scaled[sample_indices_dbscan]

# Tính k-nearest neighbors distances
k = 5  # MinPts - 1
neighbors = NearestNeighbors(n_neighbors=k)
neighbors.fit(X_sample_dbscan)
distances, indices = neighbors.kneighbors(X_sample_dbscan)

# Sort và plot k-distance graph
distances = np.sort(distances[:, k-1], axis=0)

# Áp dụng DBSCAN với eps và min_samples
eps = 3.0  # Khoảng cách tối đa giữa 2 điểm để được coi là neighbors
min_samples = 5  # Số điểm tối thiểu để tạo thành core point

dbscan = DBSCAN(eps=eps, min_samples=min_samples, n_jobs=-1)
dbscan_labels = dbscan.fit_predict(X_train_scaled)

# Phân tích kết quả
n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = list(dbscan_labels).count(-1)
mask = dbscan_labels != -1  # Mask cho non-noise points

print(f"\n📊 Kết quả DBSCAN Clustering:")
print(f"   Số clusters tìm được: {n_clusters}")
print(f"   Số noise points: {n_noise} ({n_noise/len(dbscan_labels)*100:.1f}%)")
print(f"   Số core points: {len(dbscan_labels) - n_noise}")

# Đánh giá (chỉ với non-noise points)
if n_clusters > 1 and mask.sum() > 0:
    sil_score = silhouette_score(X_train_scaled[mask], dbscan_labels[mask])
    ari_score = adjusted_rand_score(y_train[mask], dbscan_labels[mask])
    print(f"   Silhouette Score (non-noise): {sil_score:.4f}")
    print(f"   Adjusted Rand Index (non-noise): {ari_score:.4f}")
else:
    sil_score = 0
    ari_score = 0
    print(f"   ⚠️ Không đủ clusters để đánh giá (cần ít nhất 2 clusters)")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: DBSCAN clusters
# Tạo colormap tùy chỉnh cho noise points (màu đen)
unique_labels = set(dbscan_labels)
colors = ['#%06X' % np.random.randint(0, 0xFFFFFF) for _ in range(len(unique_labels))]

for label, color in zip(unique_labels, colors):
    if label == -1:
        color = 'black'  # Noise points màu đen
        marker = 'x'
        alpha = 0.5
        label_text = 'Noise'
    else:
        marker = 'o'
        alpha = 0.6
        label_text = f'Cluster {label}'
    
    mask_plot = dbscan_labels == label
    axes[0].scatter(X_pca[mask_plot, 0], X_pca[mask_plot, 1], 
                    c=color, marker=marker, alpha=alpha, 
                    edgecolors='black', linewidth=0.5,
                    label=label_text if label == -1 or label < 3 else '')

axes[0].set_xlabel('PC1', fontsize=12, fontweight='bold')
axes[0].set_ylabel('PC2', fontsize=12, fontweight='bold')
title = f'DBSCAN Clustering (eps={eps}, min_samples={min_samples})'
if n_clusters > 1 and mask.sum() > 0:
    title += f'\nSilhouette: {sil_score:.3f}, Clusters: {n_clusters}, Noise: {n_noise}'
else:
    title += f'\nClusters: {n_clusters}, Noise: {n_noise}'
axes[0].set_title(title, fontsize=14, fontweight='bold')
axes[0].legend(loc='best')
axes[0].grid(True, alpha=0.3)

# Plot 2: K-distance graph để chọn eps
axes[1].plot(range(len(distances)), distances, 'b-', linewidth=1)
axes[1].axhline(y=eps, color='red', linestyle='--', linewidth=2, label=f'eps={eps} (chosen)')
axes[1].set_xlabel('Points sorted by distance', fontsize=12, fontweight='bold')
axes[1].set_ylabel(f'{k}-th Nearest Neighbor Distance', fontsize=12, fontweight='bold')
axes[1].set_title('K-Distance Graph\n(Elbow → Optimal eps)', fontsize=14, fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n💡 Đặc điểm DBSCAN:")
print(f"   ✅ Ưu điểm: Tìm clusters có hình dạng bất kỳ, phát hiện outliers")
print(f"   ✅ Không cần biết trước số clusters")
print(f"   ⚠️ Nhược điểm: Khó chọn tham số eps và min_samples")
print(f"   ⚠️ Không hoạt động tốt với dữ liệu có mật độ khác nhau")

if n_noise > len(dbscan_labels) * 0.3:
    print(f"\n⚠️ CẢNH BÁO: Tỷ lệ noise cao ({n_noise/len(dbscan_labels)*100:.1f}%)")
    print(f"   → Có thể cần điều chỉnh eps (tăng) hoặc min_samples (giảm)")
```

```python
# So sánh các thuật toán clustering
print("📊 SO SÁNH CÁC THUẬT TOÁN CLUSTERING\n")
print("="*80)

# Tạo DataFrame so sánh
comparison_data = {
    'Thuật Toán': ['K-Means', 'Hierarchical', 'DBSCAN'],
    'Số Clusters': [
        optimal_k, 
        3, 
        n_clusters if 'n_clusters' in locals() else 'N/A'
    ],
    'Silhouette Score': [
        silhouette_avg,
        sil_hier,
        sil_score if 'sil_score' in locals() and n_clusters > 1 and mask.sum() > 0 else 0
    ],
    'ARI (vs True Labels)': [
        ari, 
        ari_hier,
        ari_score if 'ari_score' in locals() and n_clusters > 1 and mask.sum() > 0 else 0
    ],
    'Noise Points': [
        0, 
        0, 
        n_noise if 'n_noise' in locals() else 0
    ]
}

comparison_df = pd.DataFrame(comparison_data)

print(comparison_df.to_string(index=False))
print("="*80)

# Visualize comparison
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: Silhouette Score comparison
algorithms = comparison_df['Thuật Toán']
silhouette_values = comparison_df['Silhouette Score']
colors_bar = ['#3498db', '#2ecc71', '#e74c3c']

axes[0, 0].bar(algorithms, silhouette_values, color=colors_bar, edgecolor='black', linewidth=2)
axes[0, 0].set_ylabel('Silhouette Score', fontsize=12, fontweight='bold')
axes[0, 0].set_title('Silhouette Score Comparison\n(Higher is Better, Range: -1 to 1)', 
                  fontsize=14, fontweight='bold')
axes[0, 0].set_ylim([0, max(silhouette_values) * 1.2])
axes[0, 0].grid(True, alpha=0.3, axis='y')

# Thêm giá trị lên cột
for i, v in enumerate(silhouette_values):
    axes[0, 0].text(i, v + 0.01, f'{v:.3f}', ha='center', fontweight='bold', fontsize=11)

# Plot 2: ARI comparison
ari_values = comparison_df['ARI (vs True Labels)']
axes[0, 1].bar(algorithms, ari_values, color=colors_bar, edgecolor='black', linewidth=2)
axes[0, 1].set_ylabel('Adjusted Rand Index', fontsize=12, fontweight='bold')
axes[0, 1].set_title('ARI vs True Labels\n(Higher is Better, Range: -1 to 1)', 
                  fontsize=14, fontweight='bold')
axes[0, 1].set_ylim([min(0, min(ari_values) * 1.2), max(ari_values) * 1.2])
axes[0, 1].grid(True, alpha=0.3, axis='y')

# Thêm giá trị lên cột
for i, v in enumerate(ari_values):
    axes[0, 1].text(i, v + 0.01, f'{v:.3f}', ha='center', fontweight='bold', fontsize=11)

# Plot 3: Noise/Outlier Detection
noise_counts = comparison_df['Noise Points']
axes[1, 0].bar(algorithms, noise_counts, color=colors_bar, edgecolor='black', linewidth=2)
axes[1, 0].set_ylabel('Number of Noise Points', fontsize=12, fontweight='bold')
axes[1, 0].set_title('Outlier Detection\n(DBSCAN can identify noise points)', 
                  fontsize=14, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3, axis='y')

# Thêm giá trị lên cột
for i, v in enumerate(noise_counts):
    if v > 0:
        axes[1, 0].text(i, v + max(noise_counts)*0.02, f'{v}', ha='center', fontweight='bold', fontsize=11)

# Plot 4: All clusters visualization in one plot
# Sử dụng 3 subplots nhỏ trong axes[1,1]
ax_main = axes[1, 1]
ax_main.axis('off')

# Create 3 mini subplots
from matplotlib.gridspec import GridSpec
gs = GridSpec(1, 3, figure=fig, left=0.525, right=0.975, bottom=0.1, top=0.45, wspace=0.3)
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[0, 2])

# K-Means
ax1.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, cmap='viridis', 
            alpha=0.5, s=10, edgecolors='none')
ax1.set_title('K-Means', fontsize=10, fontweight='bold')
ax1.set_xlabel('PC1', fontsize=8)
ax1.set_ylabel('PC2', fontsize=8)
ax1.grid(True, alpha=0.3)

# Hierarchical
ax2.scatter(X_pca[:, 0], X_pca[:, 1], c=hier_labels_full, cmap='viridis', 
            alpha=0.5, s=10, edgecolors='none')
ax2.set_title('Hierarchical', fontsize=10, fontweight='bold')
ax2.set_xlabel('PC1', fontsize=8)
ax2.set_ylabel('PC2', fontsize=8)
ax2.grid(True, alpha=0.3)

# DBSCAN
if 'dbscan_labels' in locals():
    scatter = ax3.scatter(X_pca[:, 0], X_pca[:, 1], c=dbscan_labels, cmap='viridis', 
                alpha=0.5, s=10, edgecolors='none')
    # Highlight noise points
    noise_mask = dbscan_labels == -1
    ax3.scatter(X_pca[noise_mask, 0], X_pca[noise_mask, 1], c='black', 
                marker='x', s=20, alpha=0.8, label='Noise')
    ax3.legend(fontsize=6)
ax3.set_title('DBSCAN', fontsize=10, fontweight='bold')
ax3.set_xlabel('PC1', fontsize=8)
ax3.set_ylabel('PC2', fontsize=8)
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Phân tích chi tiết
print(f"\n" + "="*80)
print(f"📈 PHÂN TÍCH CHI TIẾT")
print(f"="*80)

# Tìm thuật toán tốt nhất theo từng tiêu chí
best_sil = comparison_df.loc[comparison_df['Silhouette Score'].idxmax()]
best_ari = comparison_df.loc[comparison_df['ARI (vs True Labels)'].idxmax()]

print(f"\n🥇 THUẬT TOÁN TỐT NHẤT:")
print(f"   • Về Silhouette Score: {best_sil['Thuật Toán']} ({best_sil['Silhouette Score']:.4f})")
print(f"   • Về ARI: {best_ari['Thuật Toán']} ({best_ari['ARI (vs True Labels)']:.4f})")

print(f"\n📊 DIỄN GIẢI CÁC CHỈ SỐ:")
print(f"   • Silhouette Score:")
print(f"     - Gần 1: Clusters rất tách biệt và compact (tốt)")
print(f"     - Gần 0: Clusters chồng lấn nhau (trung bình)")
print(f"     - Âm: Điểm được gán sai cluster (xấu)")
print(f"\n   • Adjusted Rand Index (ARI):")
print(f"     - 1.0: Phân cụm hoàn hảo khớp với nhãn thực tế")
print(f"     - 0.0: Phân cụm ngẫu nhiên")
print(f"     - Âm: Phân cụm tệ hơn ngẫu nhiên")

print(f"\n🔍 PHÂN TÍCH TỪNG THUẬT TOÁN:")
print(f"\n   1️⃣ K-MEANS:")
print(f"      ✅ Ưu điểm: Nhanh, đơn giản, scalable")
print(f"      ✅ Tốt cho: Clusters hình cầu, kích thước đồng đều")
print(f"      ⚠️ Hạn chế: Cần biết trước số clusters (k)")
print(f"      📊 Kết quả: Sil={silhouette_avg:.3f}, ARI={ari:.3f}")

print(f"\n   2️⃣ HIERARCHICAL CLUSTERING:")
print(f"      ✅ Ưu điểm: Có dendrogram, không cần init random")
print(f"      ✅ Tốt cho: Phân tích cấu trúc phân cấp dữ liệu")
print(f"      ⚠️ Hạn chế: Chậm O(n²), tốn bộ nhớ")
print(f"      📊 Kết quả: Sil={sil_hier:.3f}, ARI={ari_hier:.3f}")

if 'n_clusters' in locals() and n_clusters > 0:
    print(f"\n   3️⃣ DBSCAN:")
    print(f"      ✅ Ưu điểm: Tự động tìm số clusters, phát hiện outliers")
    print(f"      ✅ Tốt cho: Clusters hình dạng bất kỳ, có nhiễu")
    print(f"      ⚠️ Hạn chế: Khó chọn tham số (eps, min_samples)")
    print(f"      📊 Kết quả: {n_clusters} clusters, {n_noise} noise points")
    if n_clusters > 1 and mask.sum() > 0:
        print(f"      📊 Đánh giá: Sil={sil_score:.3f}, ARI={ari_score:.3f}")
else:
    print(f"\n   3️⃣ DBSCAN:")
    print(f"      ⚠️ Chưa chạy hoặc không tìm được clusters phù hợp")

print(f"\n" + "="*80)
print(f"💡 KẾT LUẬN VÀ KHUYẾN NGHỊ:")
print(f"="*80)

# Đưa ra khuyến nghị dựa trên kết quả
if silhouette_avg > 0.5 or ari > 0.5:
    print(f"   ✅ Dữ liệu có cấu trúc clustering tốt!")
    print(f"   ✅ Các class có thể phân biệt rõ ràng bằng các đặc trưng trích xuất")
else:
    print(f"   ⚠️ Clustering scores thấp:")
    print(f"   → Các class có thể chồng lấn về mặt đặc trưng")
    print(f"   → Cần thêm feature engineering hoặc deep features")

print(f"\n   🎯 Lựa chọn thuật toán:")
print(f"   • Dữ liệu lớn, cần nhanh → K-Means")
print(f"   • Cần hiểu cấu trúc phân cấp → Hierarchical")
print(f"   • Có nhiễu, không biết số clusters → DBSCAN")
print(f"   • Dữ liệu phức tạp → Thử GMM (Gaussian Mixture Model)")

print(f"\n" + "="*80)
```

```python
# Định nghĩa các thuật toán classification
print("🔄 Huấn luyện và đánh giá nhiều thuật toán classification...")
print("   Đây có thể mất vài phút...\n")

classifiers = {
    'Decision Tree': DecisionTreeClassifier(max_depth=15, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1),
    'SVM (RBF)': SVC(kernel='rbf', random_state=42, probability=True),
    'Naive Bayes': GaussianNB(),
    'k-NN (k=5)': KNeighborsClassifier(n_neighbors=5, n_jobs=-1)
}

# Lưu kết quả
results = []
trained_models = {}

for name, clf in classifiers.items():
    print(f"📊 Training {name}...", end=" ")
    
    # Train model
    clf.fit(X_train_scaled, y_train)
    trained_models[name] = clf
    
    # Predictions
    y_train_pred = clf.predict(X_train_scaled)
    y_val_pred = clf.predict(X_val_scaled)
    
    # Calculate metrics
    train_acc = accuracy_score(y_train, y_train_pred)
    val_acc = accuracy_score(y_val, y_val_pred)
    val_precision = precision_score(y_val, y_val_pred, average='weighted', zero_division=0)
    val_recall = recall_score(y_val, y_val_pred, average='weighted', zero_division=0)
    val_f1 = f1_score(y_val, y_val_pred, average='weighted', zero_division=0)
    
    # Cross-validation score
    cv_scores = cross_val_score(clf, X_train_scaled, y_train, cv=5, scoring='accuracy', n_jobs=-1)
    
    results.append({
        'Algorithm': name,
        'Train Accuracy': train_acc,
        'Val Accuracy': val_acc,
        'Precision': val_precision,
        'Recall': val_recall,
        'F1-Score': val_f1,
        'CV Mean': cv_scores.mean(),
        'CV Std': cv_scores.std()
    })
    
    print(f"✓ Val Acc: {val_acc:.4f}, F1: {val_f1:.4f}")

# Tạo DataFrame kết quả
results_df = pd.DataFrame(results).sort_values('Val Accuracy', ascending=False)

print("\n" + "="*90)
print("📊 KẾT QUẢ SO SÁNH CÁC THUẬT TOÁN CLASSIFICATION")
print("="*90)
print(results_df.to_string(index=False))
print("="*90)

# Tìm thuật toán tốt nhất
best_model_name = results_df.iloc[0]['Algorithm']
best_val_acc = results_df.iloc[0]['Val Accuracy']

print(f"\n🏆 THUẬT TOÁN TỐT NHẤT: {best_model_name}")
print(f"   Validation Accuracy: {best_val_acc:.4f}")
print(f"   F1-Score: {results_df.iloc[0]['F1-Score']:.4f}")
```

```python
# Tạo các ensemble models
print("🔄 Huấn luyện Ensemble Methods...")
print("   Đây có thể mất 5-10 phút...\n")

ensemble_models = {}
ensemble_results = []

# 1. Voting Classifier (Soft Voting)
print("📊 Training Voting Classifier...", end=" ")
voting_clf = VotingClassifier(
    estimators=[
        ('rf', RandomForestClassifier(n_estimators=50, max_depth=15, random_state=42)),
        ('svm', SVC(kernel='rbf', probability=True, random_state=42)),
        ('knn', KNeighborsClassifier(n_neighbors=5))
    ],
    voting='soft',
    n_jobs=-1
)
voting_clf.fit(X_train_scaled, y_train)
ensemble_models['Voting (Soft)'] = voting_clf

y_val_pred = voting_clf.predict(X_val_scaled)
val_acc = accuracy_score(y_val, y_val_pred)
val_f1 = f1_score(y_val, y_val_pred, average='weighted')
ensemble_results.append({
    'Model': 'Voting Classifier',
    'Val Accuracy': val_acc,
    'F1-Score': val_f1
})
print(f"✓ Val Acc: {val_acc:.4f}")

# 2. Bagging
print("📊 Training Bagging Classifier...", end=" ")
bagging_clf = BaggingClassifier(
    estimator=DecisionTreeClassifier(max_depth=15, random_state=42),
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)
bagging_clf.fit(X_train_scaled, y_train)
ensemble_models['Bagging'] = bagging_clf

y_val_pred = bagging_clf.predict(X_val_scaled)
val_acc = accuracy_score(y_val, y_val_pred)
val_f1 = f1_score(y_val, y_val_pred, average='weighted')
ensemble_results.append({
    'Model': 'Bagging',
    'Val Accuracy': val_acc,
    'F1-Score': val_f1
})
print(f"✓ Val Acc: {val_acc:.4f}")

# 3. AdaBoost
print("📊 Training AdaBoost...", end=" ")
ada_clf = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=5, random_state=42),
    n_estimators=50,
    random_state=42
)
ada_clf.fit(X_train_scaled, y_train)
ensemble_models['AdaBoost'] = ada_clf

y_val_pred = ada_clf.predict(X_val_scaled)
val_acc = accuracy_score(y_val, y_val_pred)
val_f1 = f1_score(y_val, y_val_pred, average='weighted')
ensemble_results.append({
    'Model': 'AdaBoost',
    'Val Accuracy': val_acc,
    'F1-Score': val_f1
})
print(f"✓ Val Acc: {val_acc:.4f}")

# 4. Gradient Boosting
print("📊 Training Gradient Boosting...", end=" ")
gb_clf = GradientBoostingClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42
)
gb_clf.fit(X_train_scaled, y_train)
ensemble_models['Gradient Boosting'] = gb_clf

y_val_pred = gb_clf.predict(X_val_scaled)
val_acc = accuracy_score(y_val, y_val_pred)
val_f1 = f1_score(y_val, y_val_pred, average='weighted')
ensemble_results.append({
    'Model': 'Gradient Boosting',
    'Val Accuracy': val_acc,
    'F1-Score': val_f1
})
print(f"✓ Val Acc: {val_acc:.4f}")

# So sánh kết quả
print("\n" + "="*70)
print("📊 SO SÁNH CÁC ENSEMBLE METHODS")
print("="*70)
ensemble_df = pd.DataFrame(ensemble_results).sort_values('Val Accuracy', ascending=False)
print(ensemble_df.to_string(index=False))
print("="*70)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar plot cho Val Accuracy
ax1 = axes[0]
colors_ensemble = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A']
bars = ax1.barh(ensemble_df['Model'], ensemble_df['Val Accuracy'], color=colors_ensemble)
ax1.set_xlabel('Validation Accuracy', fontsize=12)
ax1.set_title('🎯 Validation Accuracy Comparison', fontsize=14, fontweight='bold')
ax1.set_xlim([0, 1])

# Thêm giá trị lên mỗi thanh
for i, (bar, acc) in enumerate(zip(bars, ensemble_df['Val Accuracy'])):
    ax1.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2, 
             f'{acc:.4f}', va='center', fontweight='bold')

# Bar plot cho F1-Score
ax2 = axes[1]
bars2 = ax2.barh(ensemble_df['Model'], ensemble_df['F1-Score'], color=colors_ensemble)
ax2.set_xlabel('F1-Score', fontsize=12)
ax2.set_title('📊 F1-Score Comparison', fontsize=14, fontweight='bold')
ax2.set_xlim([0, 1])

# Thêm giá trị lên mỗi thanh
for i, (bar, f1) in enumerate(zip(bars2, ensemble_df['F1-Score'])):
    ax2.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2, 
             f'{f1:.4f}', va='center', fontweight='bold')

plt.tight_layout()
plt.show()

# Tìm mô hình tốt nhất
best_ensemble = ensemble_df.iloc[0]
print(f"\n🏆 ENSEMBLE TỐT NHẤT: {best_ensemble['Model']}")
print(f"   Validation Accuracy: {best_ensemble['Val Accuracy']:.4f}")
print(f"   F1-Score: {best_ensemble['F1-Score']:.4f}")
```

```python
# 3. ROC Curves cho từng class
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import roc_curve, auc

# Get probability predictions from the best model
best_model = trained_models[best_model_name]
y_val_pred_proba = best_model.predict_proba(X_val_scaled)

# Chuyển labels sang binary format cho ROC
le = LabelEncoder()
y_val_encoded = le.fit_transform(y_val)
y_val_binarized = label_binarize(y_val_encoded, classes=[0, 1, 2])
n_classes = y_val_binarized.shape[1]

# Tính ROC curve và AUC cho từng class
fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_val_binarized[:, i], y_val_pred_proba[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Vẽ ROC curves
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
class_names_display = ['Angular Leaf Spot', 'Bean Rust', 'Healthy']

for i, (ax, color, class_name) in enumerate(zip(axes, colors, class_names_display)):
    ax.plot(fpr[i], tpr[i], color=color, lw=3,
            label=f'ROC curve (AUC = {roc_auc[i]:.3f})')
    ax.plot([0, 1], [0, 1], 'k--', lw=2, label='Random Classifier')
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.05])
    ax.set_xlabel('False Positive Rate', fontsize=11, weight='bold')
    ax.set_ylabel('True Positive Rate', fontsize=11, weight='bold')
    ax.set_title(f'ROC: {class_name}', fontsize=13, weight='bold', pad=10)
    ax.legend(loc="lower right", fontsize=10)
    ax.grid(alpha=0.3)
    
    # Tô màu vùng dưới đường cong
    ax.fill_between(fpr[i], tpr[i], alpha=0.2, color=color)

plt.suptitle(f'🎯 ROC Curves - {best_model_name}', fontsize=16, weight='bold', y=1.02)
plt.tight_layout()
plt.show()

# In AUC scores
print("\n" + "="*70)
print("📊 AUC SCORES PER CLASS")
print("="*70)
for i, class_name in enumerate(class_names_display):
    print(f"   {class_name:20s}: AUC = {roc_auc[i]:.4f}")
print(f"\n   Mean AUC: {np.mean(list(roc_auc.values())):.4f}")
print("="*70)

print("\n💡 GIẢI THÍCH ROC-AUC:")
print("   ✓ AUC càng gần 1.0 → Model phân loại càng tốt")
print("   ✓ AUC = 0.5 → Model random (không học được gì)")
print("   ✓ AUC > 0.9 → Model rất tốt")
print("   ✓ AUC > 0.95 → Model xuất sắc")
```

```python
def train_one_epoch(model, dataloader, criterion, optimizer, device):
    """Train the model for one epoch."""
    model.train()
    running_loss = 0.0
    running_corrects = 0
    
    # Progress bar
    pbar = tqdm(dataloader, desc='Training', leave=False)
    
    for inputs, labels in pbar:
        inputs, labels = inputs.to(device), labels.to(device).long()
        
        # Zero gradients
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        _, preds = torch.max(outputs, 1)
        
        # Backward pass
        loss.backward()
        optimizer.step()
        
        # Statistics
        running_loss += loss.item() * inputs.size(0)
        running_corrects += torch.sum(preds == labels.data)
        
        # Update progress bar
        pbar.set_postfix({'loss': f'{loss.item():.4f}'})
    
    epoch_loss = running_loss / len(dataloader.dataset)
    epoch_acc = running_corrects.double() / len(dataloader.dataset)
    
    return epoch_loss, epoch_acc.item()


def validate(model, dataloader, criterion, device):
    """Validate the model."""
    model.eval()
    running_loss = 0.0
    running_corrects = 0
    
    all_preds = []
    all_labels = []
    
    # Progress bar
    pbar = tqdm(dataloader, desc='Validation', leave=False)
    
    with torch.no_grad():
        for inputs, labels in pbar:
            inputs, labels = inputs.to(device), labels.to(device).long()
            
            # Forward pass
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            _, preds = torch.max(outputs, 1)
            
            # Statistics
            running_loss += loss.item() * inputs.size(0)
            running_corrects += torch.sum(preds == labels.data)
            
            # Store for confusion matrix
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            
            # Update progress bar
            pbar.set_postfix({'loss': f'{loss.item():.4f}'})
    
    epoch_loss = running_loss / len(dataloader.dataset)
    epoch_acc = running_corrects.double() / len(dataloader.dataset)
    
    return epoch_loss, epoch_acc.item(), all_preds, all_labels


def train_model(model, train_loader, val_loader, criterion, optimizer, scheduler, 
                num_epochs=20, patience=7, save_path='best_model.pth'):
    """
    Complete training loop with early stopping.
    
    Args:
        model: PyTorch model
        train_loader: Training data loader
        val_loader: Validation data loader
        criterion: Loss function
        optimizer: Optimizer
        scheduler: Learning rate scheduler
        num_epochs: Maximum number of epochs
        patience: Early stopping patience
        save_path: Path to save best model
    
    Returns:
        history: Dictionary containing training history
        best_model: Best model state dict
    """
    
    history = {
        'train_loss': [],
        'train_acc': [],
        'val_loss': [],
        'val_acc': [],
        'lr': []
    }
    
    best_val_acc = 0.0
    best_model_wts = None
    epochs_no_improve = 0
    
    print("\n" + "="*60)
    print("STARTING TRAINING")
    print("="*60)
    
    start_time = time.time()
    
    for epoch in range(num_epochs):
        print(f'\nEpoch {epoch + 1}/{num_epochs}')
        print('-' * 40)
        
        # Training phase
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
        
        # Validation phase
        val_loss, val_acc, _, _ = validate(model, val_loader, criterion, device)
        
        # Update learning rate
        scheduler.step(val_loss)
        
        # Get current learning rate
        current_lr = optimizer.param_groups[0]['lr']
        
        # Store metrics
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        history['lr'].append(current_lr)
        
        # Print epoch results
        print(f'Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.4f}')
        print(f'Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.4f}')
        print(f'Learning Rate: {current_lr:.6f}')
        
        # Check if best model
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_model_wts = model.state_dict().copy()
            epochs_no_improve = 0
            print(f'🎯 New best model! Validation Accuracy: {val_acc:.4f}')
            
            # Save best model
            torch.save({
                'epoch': epoch,
                'model_state_dict': best_model_wts,
                'optimizer_state_dict': optimizer.state_dict(),
                'val_acc': val_acc,
                'val_loss': val_loss,
            }, save_path)
        else:
            epochs_no_improve += 1
            
        # Early stopping
        if epochs_no_improve >= patience:
            print(f'\n⏹️  Early stopping triggered after {epoch + 1} epochs')
            break
    
    # Training complete
    total_time = time.time() - start_time
    print("\n" + "="*60)
    print("TRAINING COMPLETE")
    print("="*60)
    print(f'Total time: {total_time // 60:.0f}m {total_time % 60:.0f}s')
    print(f'Best validation accuracy: {best_val_acc:.4f}')
    
    # Load best model weights
    if best_model_wts is not None:
        model.load_state_dict(best_model_wts)
    
    return history, model

print("✅ Training functions defined successfully!")
```

# **Bước 3: Huấn Luyện và Kiểm Định Mô Hình**
Phần này sẽ bao gồm vòng lặp huấn luyện và các hàm cần thiết cho huấn luyện (train) và kiểm định (test). Nó sẽ chi tiết các bước của quá trình huấn luyện, bao gồm thiết lập hàm mất mát, optimizer, learning rate scheduler, và chính vòng lặp huấn luyện, bao gồm cả early stopping.

### **3.1 Nhập Các Thư Viện Cần Thiết**
Đầu tiên, chúng ta cần đảm bảo rằng tất cả các thư viện cần thiết đã được nhập.


```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim import lr_scheduler
from tqdm import tqdm
```

### **3.2 Định Nghĩa Hàm Mất Mát, Optimizer và Learning Rate Scheduler**
Trước khi huấn luyện, chúng ta định nghĩa hàm mất mát, optimizer và learning rate scheduler.

```python
# Hàm mất mát
criterion = nn.CrossEntropyLoss()

# Optimizer (sử dụng AdamW ở đây)
optimizer = optim.AdamW(model.parameters(), lr=0.0001)

# Learning rate scheduler (tùy chọn)
lr_scheduler = lr_scheduler.StepLR(optimizer, step_size=7, gamma=0.1)
```

### **3.3 Hàm Huấn Luyện và Kiểm Định**
Định nghĩa các hàm để huấn luyện và kiểm định mô hình.

```python
def train_model(model, criterion, optimizer, scheduler, num_epochs=25):
    train_loss_history = []
    train_acc_history = []
    val_loss_history = []
    val_acc_history = []

    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}')
        print('-' * 10)

        # Mỗi epoch có giai đoạn huấn luyện và kiểm định
        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()  # Đặt mô hình ở chế độ huấn luyện
                dataloader = train_loader
            else:
                model.eval()   # Đặt mô hình ở chế độ đánh giá
                dataloader = val_loader

            running_loss = 0.0
            running_corrects = 0

            # Lặp qua dữ liệu
            for inputs, labels in dataloader:
                inputs = inputs.to(device)
                labels = labels.to(device).long()

                # Đặt gradients về 0
                optimizer.zero_grad()

                # Forward pass
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    # Backward + optimize chỉ trong giai đoạn huấn luyện
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                # Thống kê
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            if phase == 'train':
                scheduler.step()

            epoch_loss = running_loss / len(dataloader.dataset)
            epoch_acc = running_corrects.double() / len(dataloader.dataset)

            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            # Lưu các metrics dưới dạng số Python
            if phase == 'train':
                train_loss_history.append(epoch_loss)
                train_acc_history.append(epoch_acc.item())  # Chuyển sang số Python
            else:
                val_loss_history.append(epoch_loss)
                val_acc_history.append(epoch_acc.item())  # Chuyển sang số Python

    print('Huấn luyện hoàn tất')

    return model, train_loss_history, train_acc_history, val_loss_history, val_acc_history
```

### **3.4 Huấn Luyện Mô Hình**

### **4.1 Đánh Giá Mô Hình**
Đầu tiên, chúng ta sẽ tạo một hàm để đánh giá mô hình. Hàm này sẽ sử dụng tập dữ liệu kiểm định để dự đoán nhãn và sau đó so sánh chúng với nhãn thực tế.

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import seaborn as sns

def evaluate_model(model, dataloader):
    model.eval()  # Đặt mô hình ở chế độ đánh giá
    all_preds = []
    all_labels = []

    with torch.no_grad():  # Không cần theo dõi gradients
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    return all_labels, all_preds

# Đánh giá mô hình
all_labels, all_preds = evaluate_model(model, val_loader)
```

### **4.2 Báo Cáo Phân Loại và Độ Chính Xác**
Chúng ta sẽ tạo báo cáo phân loại và tính toán độ chính xác.

```python
print("Báo Cáo Phân Loại:")
print(classification_report(all_labels, all_preds, target_names=["healthy", "angular_leaf_spot", "bean_rust"]))

accuracy = accuracy_score(all_labels, all_preds)
print(f"Độ chính xác: {accuracy * 100:.2f}%")
```

Bước này cung cấp cái nhìn chi tiết về hiệu suất của mô hình trên tập kiểm định, làm nổi bật điểm mạnh và các lĩnh vực có thể cần cải thiện. Báo cáo phân loại đưa ra cái nhìn tổng quan toàn diện về precision, recall và F1-score của mô hình cho từng lớp, trong khi ma trận nhầm lẫn trực quan hóa các dự đoán của mô hình so với nhãn thực tế.

### **5.1 Hiển Thị Dự Đoán Chính Xác**
Tiếp theo, hãy trực quan hóa một instance từ tập dữ liệu kiểm định nơi mô hình của chúng ta đã dự đoán chính xác. Chúng ta sẽ hiển thị ảnh, nhãn thực tế và nhãn dự đoán.

```python
import matplotlib.pyplot as plt
import torch.nn.functional as F

def visualize_multiple_predictions(dataloader, model, device, num_images=15):
    model.eval()
    images_shown = 0
    plt.figure(figsize=(15, 15))  # Tăng kích thước figure để có khoảng cách tốt hơn

    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            probabilities = F.softmax(outputs, dim=1)
            _, preds = torch.max(outputs, 1)

            for i in range(inputs.shape[0]):
                if images_shown == num_images:
                    break

                confidence = torch.max(probabilities[i]).item() * 100  # Chuyển sang phần trăm
                img = inputs[i].cpu().data.numpy().transpose((1, 2, 0))
                mean = np.array([0.485, 0.456, 0.406])
                std = np.array([0.229, 0.224, 0.225])
                img = std * img + mean
                img = np.clip(img, 0, 1)

                ax = plt.subplot(num_images // 5, 5, images_shown + 1)
                ax.imshow(img)
                ax.set_title(f'Dự đoán: {preds[i].item()}, Thực tế: {labels[i].item()}\nĐộ tin cậy: {confidence:.2f}%', fontsize=10)
                ax.axis('off')

                images_shown += 1

            if images_shown == num_images:
                break

    plt.subplots_adjust(hspace=0.03)  # Điều chỉnh khoảng cách ngang
    plt.show()
```

```python
# Gọi hàm để trực quan hóa dự đoán
visualize_multiple_predictions(val_loader, model, device, num_images=15)
```
